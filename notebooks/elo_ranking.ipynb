# 🏆 Elo Ranking System

This notebook demonstrates how to build a dynamic ranking system using the **Elo Rating Algorithm**.

We will:
- Initialize player ratings
- Calculate expected win probabilities
- Update ratings after matches
- Process multiple historical matches
- Generate a final leaderboard


````python
# Configuration
INITIAL_RATING = 1500
K_FACTOR = 32
````

## 🧠 1. Expected Score

The expected probability of Player A winning is:

`EA = 1 / (1 + 10 ** ((RB - RA) / 400))`


````python
def expected_score(rating_a, rating_b):
    """Calculate Player A's expected probability of winning."""
    return 1 / (1 + 10 ** ((rating_b - rating_a) / 400))


# Example
player_a = 1500
player_b = 1600

print(f'Player A expected score: {expected_score(player_a, player_b):.3f}')
print(f'Player B expected score: {expected_score(player_b, player_a):.3f}')
````

## ⚡ 2. Update Elo Ratings

Formula:

`New Rating = Old Rating + K × (Actual Score - Expected Score)`

Results:
- `1` = Player A wins
- `0.5` = Draw
- `0` = Player B wins


````python
def update_ratings(rating_a, rating_b, score_a, k=K_FACTOR):
    """Update Elo ratings after a match.

    score_a: 1 = A wins, 0.5 = draw, 0 = B wins
    """
    expected_a = expected_score(rating_a, rating_b)
    expected_b = expected_score(rating_b, rating_a)

    score_b = 1 - score_a

    new_rating_a = rating_a + k * (score_a - expected_a)
    new_rating_b = rating_b + k * (score_b - expected_b)

    return new_rating_a, new_rating_b
````

## 🥊 3. Single Match Example

````python
player_a_rating = 1500
player_b_rating = 1500

new_a, new_b = update_ratings(
    player_a_rating,
    player_b_rating,
    score_a=1
)

print(f'Player A: {player_a_rating} → {new_a:.2f}')
print(f'Player B: {player_b_rating} → {new_b:.2f}')
````

## 📚 4. Process Multiple Matches

````python
matches = [
    ('Alice', 'Bob', 1),      # Alice wins
    ('Bob', 'Charlie', 1),    # Bob wins
    ('Alice', 'Charlie', 0),  # Charlie wins
    ('Alice', 'Bob', 0.5),    # Draw
    ('Charlie', 'Bob', 1),    # Charlie wins
]
````

````python
ratings = {}

def get_rating(player):
    if player not in ratings:
        ratings[player] = INITIAL_RATING
    return ratings[player]


for player_a, player_b, result in matches:
    rating_a = get_rating(player_a)
    rating_b = get_rating(player_b)

    new_a, new_b = update_ratings(rating_a, rating_b, result)

    ratings[player_a] = new_a
    ratings[player_b] = new_b

    print(f'{player_a} vs {player_b} | Result: {result}')
    print(f'  {player_a}: {new_a:.2f}')
    print(f'  {player_b}: {new_b:.2f}\n')
````

## 🏅 5. Generate Final Rankings

````python
ranking = sorted(ratings.items(), key=lambda x: x[1], reverse=True)

print('🏆 FINAL LEADERBOARD')
print('-' * 35)

for rank, (player, rating) in enumerate(ranking, start=1):
    print(f'{rank}. {player:<10} {rating:.2f}')
````

## 📊 6. Ranking System Workflow

```text
Match Data
    ↓
Initialize Ratings
    ↓
Calculate Expected Score
    ↓
Compare Actual vs Expected Result
    ↓
Update Elo Ratings
    ↓
Store Ratings
    ↓
Generate Leaderboard
```
